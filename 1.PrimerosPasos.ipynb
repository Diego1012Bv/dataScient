## Funciones

```python
def clasificarEdad(edad):
    """ Recibe una edad y regresa la categoria"""
    if edad<12:
        return "niño"
    elif edad<18:
        return "adolescente"
    elif edad<65:
        return "adulto"
    else:
        return "adulto mayor"
```

```python
for edad in [5,15,40,80]:
    print(edad,"->",clasificarEdad(edad))
```
